```python
import pandas as pd
import numpy as np
```

```python
data = [
    ['sunny','hot','high','weak','yes'],
    ['sunny','hot','high','strong','no'],
    ['rain','cool','normal','strong','no'],
    ['sunny','mild','high','weak','yes'],
]

columns = ['outlook' , 'temparature' , 'humidity' , 'wind' , 'play tennis ']

df = pd.DataFrame(data , columns = columns)
```

```python
df.shape
```

```python
indx =0
for i in range(df.shape[0]):
  if df.iloc[i, 4] =='yes':
    indx =i
    break
```

```python
print(indx)
```

```python
state =[]
for j in range(df.shape[1]):
  state.append(df.iloc[indx,j])
```

```python
state
```

```python
new_state = list(state[:-1])
hypothesis_count = 1

for k in range(df.shape[0]):
  if df.iloc[k]['play tennis '] == 'yes':
    current_example_features = list(df.iloc[k, :-1])

    for j in range(len(new_state)):
      if new_state[j] != current_example_features[j]:
        new_state[j] = '?'

    print(f"h{hypothesis_count}: {new_state}")
    hypothesis_count += 1
```

```python
print(new_state)
```

```python
data_new = [
    ['sunny', 'warm', 'normal', 'strong', 'warm', 'same', 'yes'],
    ['sunny', 'warm', 'high', 'strong', 'warm', 'same', 'yes'],
    ['cloudy', 'cold', 'normal', 'strong', 'warm', 'change', 'yes'],
    ['rainy', 'cold', 'high', 'weak', 'warm', 'change', 'no'],
    ['sunny', 'warm', 'high', 'weak', 'cold', 'same', 'yes'],
    ['cloudy', 'warm', 'normal', 'strong', 'cold', 'same', 'yes'],
    ['rainy', 'cold', 'normal', 'strong', 'warm', 'change', 'no'],
    ['sunny', 'mild', 'normal', 'weak', 'warm', 'same', 'yes'],
    ['cloudy', 'mild', 'high', 'strong', 'warm', 'change', 'yes'],
    ['rainy', 'mild', 'high', 'weak', 'cold', 'change', 'no'],
]

columns_new = ['sky', 'airtemp', 'humidity', 'wind', 'water', 'forecast', 'enjoysport']

df_new = pd.DataFrame(data_new, columns=columns_new)
print(df_new)
```

```python
indx_new = -1
for i in range(df_new.shape[0]):
  if df_new.iloc[i]['enjoysport'] == 'yes':
    indx_new = i
    break

if indx_new == -1:
  print("No positive examples found in df_new to initialize the hypothesis.")
else:
  initial_hypothesis_new = list(df_new.iloc[indx_new, :-1])
  specific_hypothesis = list(initial_hypothesis_new)

  hypothesis_count_new = 1
  print(f"h{hypothesis_count_new}: {specific_hypothesis}")
  hypothesis_count_new += 1

  for k in range(df_new.shape[0]):
    if df_new.iloc[k]['enjoysport'] == 'yes':
      current_example_features_new = list(df_new.iloc[k, :-1])

      for j in range(len(specific_hypothesis)):
        if specific_hypothesis[j] != current_example_features_new[j]:
          specific_hypothesis[j] = '?'

      if k != indx_new:
        print(f"h{hypothesis_count_new}: {specific_hypothesis}")
        hypothesis_count_new += 1

  print(f"\nThe most specific hypothesis for df_new is: {specific_hypothesis}")
```
